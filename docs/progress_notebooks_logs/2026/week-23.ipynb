```python
import os
import numpy as np
import pandas as pd
import seaborn as sns
from pathlib import Path
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.patches import Rectangle
from matplotlib.legend_handler import HandlerBase
from matplotlib.ticker import MaxNLocator

try:
    from IPython.display import display
except ImportError:
    def display(obj):
        if hasattr(obj, "to_string"):
            print(obj.to_string())
        else:
            print(obj)


# =========================
# Style
# =========================
sns.set_theme(context="paper", style="white")


# =========================
# Configuration
# =========================

RESULTS_DIR_ORIGINAL = Path(
    "../../../src/matlab/data/result_wlnm_original_sweep_train_ratios_10-90_pseudo_properties/prediction_scores_logs"
)

RESULTS_DIR_DIRECTED = Path(
    "../../../src/matlab/data/result_wlnm_directed_sweep_train_ratios_10-90_pseudo_properties/prediction_scores_logs"
)

RESULTS_DIR_DIR_NEG = Path(
    "../../../src/matlab/data/result_wlnm_dir_neg_sweep_train_ratios_10-90_pseudo_properties/prediction_scores_logs"
)

RESULTS_DIR_SEAL = Path(
    "../../../src/python/data/result/prediction_scores_logs"
)

FOODWEB_CSV = Path(
    "../../../src/matlab/data/foodwebs_mat/foodweb_metrics_ecosystem.csv"
)

METRICS = [
    "ROC_AUC",
    "PR_AUC",
    "Precision",
    "Recall",
    "F1Score",
    "PseudoTSS",
    "PseudoTPR",
]

METRIC_LABELS = {
    "ROC_AUC": "ROC-AUC",
    "PR_AUC": "PR-AUC",
    "Precision": "Precision",
    "Recall": "Recall",
    "F1Score": "F1",
    "PseudoTSS": "TSS",
    "PseudoTPR": "TPR",
}

# =========================
# Main comparison settings
# =========================

TRAIN_RATIO_TARGET = 90

# SEAL train ratios are often realised ratios rather than exactly 90.
# Examples: 88.8889, 90.9091, 91.1765, 92.8571, 94.1176.
# A ±5 window keeps approximate 90% SEAL runs while excluding clearly different
# ratios such as 50%, 75%, 80%, or 83.33%.
TRAIN_RATIO_TOL = 10.0

# Set this to 40 if you want the main manuscript-style node-filtered comparison.
NODES_THRESHOLD = 0

# Keep None unless you want to compare only a specific K.
K_TARGET = None

# Recommended when SEAL has fewer completed food webs than WLNM.
# True means only food webs available in every version are compared.
COMPARE_ONLY_COMPLETE_CASES = True

# Usually False because the audit block gives a compact skipped-file summary.
PRINT_SKIPPED_FILES = False

RUN_SEAL_AUDIT = True
MAX_AUDIT_SKIPPED_PRINT = 40

VERSION_ORDER = [
    "Original",
    "Directed",
    "Directed-Negative",
    "SEAL",
]

GLOB_PATTERN_ORIGINAL = "*_tax_mass_results_random_wlnm_original.csv"
GLOB_PATTERN_DIRECTED = "*_tax_mass_results_random_wlnm_directed.csv"
GLOB_PATTERN_DIR_NEG = "*_tax_mass_results_random_wlnm_dir_neg.csv"
GLOB_PATTERN_SEAL = "*_tax_mass_results_SEAL.csv"

PALETTE = {
    "Original": "#7570b3",
    "Directed": "#1f78b4",
    "Directed-Negative": "#1b9e77",
    "SEAL": "#d95f02",
}


# =========================
# Helper functions
# =========================

def normalize_fw_label(name: str) -> str:
    """
    Normalise food-web labels so filenames and metadata can be matched.

    Examples:
      Alford lake -> Alford_lake
      Alford_lake -> Alford_lake
      AEW01_tax_mass_results_SEAL -> AEW01
    """
    name = str(name).strip()
    name, _ = os.path.splitext(name)
    name = name.replace(" ", "_")

    for token in [
        "_tax_mass_results_SEAL",
        "_tax_mass_results_random_wlnm_original",
        "_tax_mass_results_random_wlnm_directed",
        "_tax_mass_results_random_wlnm_dir_neg",
        "_results_SEAL",
        "_results_random_wlnm_original",
        "_results_random_wlnm_directed",
        "_results_random_wlnm_dir_neg",
        "_random_wlnm_original",
        "_random_wlnm_directed",
        "_random_wlnm_dir_neg",
        "_results_",
        "_result_",
        "_results",
        "_random",
        "_tax_mass",
        "_tax",
        "_mass",
        "_SEAL",
    ]:
        name = name.replace(token, "_")

    return "_".join(part for part in name.split("_") if part)


def parse_foodweb_from_name(fname: str) -> str:
    """
    Extract food-web name from filenames such as:
      AEW01_tax_mass_results_SEAL.csv
      German_Creek_tax_mass_results_random_wlnm_dir_neg.csv
    """
    stem = Path(fname).stem

    for token in [
        "_tax_mass_results_",
        "_results_",
        "_tax_mass_results",
        "_results",
    ]:
        if token in stem:
            return stem.split(token)[0]

    return stem


def choose_display_name(names):
    """
    Pick one readable display name for each normalised food-web key.
    Prefer names with spaces over underscore-only names when both exist.
    """
    names = sorted(set(str(x) for x in names if pd.notna(x)))

    if not names:
        return ""

    names_with_spaces = [x for x in names if " " in x]

    if names_with_spaces:
        return sorted(names_with_spaces, key=len)[0]

    return sorted(names, key=len)[0]


def coerce_numeric(df: pd.DataFrame, cols) -> pd.DataFrame:
    for c in cols:
        if c in df.columns:
            df[c] = pd.to_numeric(df[c], errors="coerce")
    return df


def filter_train_ratio(df: pd.DataFrame) -> pd.DataFrame:
    """
    Keep rows with TrainRatio close to TRAIN_RATIO_TARGET.
    This is tolerance-based because SEAL files may store realised train ratios
    such as 90.9091 rather than exactly 90.
    """
    if TRAIN_RATIO_TARGET is None:
        return df.copy()

    if "TrainRatio" not in df.columns:
        return pd.DataFrame()

    vals = pd.to_numeric(df["TrainRatio"], errors="coerce")
    return df[np.abs(vals - TRAIN_RATIO_TARGET) <= TRAIN_RATIO_TOL].copy()


def load_one_csv(csv_path: Path, version_label: str):
    df = pd.read_csv(csv_path)

    present_metrics = [m for m in METRICS if m in df.columns]
    if not present_metrics:
        if PRINT_SKIPPED_FILES:
            print(f"[SKIP] Missing all target metrics: {csv_path.name}")
        return None

    df = coerce_numeric(df, METRICS + ["K", "TrainRatio"])
    df = filter_train_ratio(df)

    if df.empty:
        if PRINT_SKIPPED_FILES:
            print(
                f"[SKIP] No rows near TrainRatio={TRAIN_RATIO_TARGET} "
                f"± {TRAIN_RATIO_TOL}: {csv_path.name}"
            )
        return None

    if K_TARGET is not None and "K" in df.columns:
        df = df[df["K"] == K_TARGET].copy()

        if df.empty:
            if PRINT_SKIPPED_FILES:
                print(f"[SKIP] No rows for K={K_TARGET}: {csv_path.name}")
            return None

    foodweb = parse_foodweb_from_name(csv_path.name)

    out = {
        "Foodweb": foodweb,
        "FW_KEY": normalize_fw_label(foodweb),
        "Version": version_label,
        "n_rows": int(len(df)),
        "file": csv_path.name,
        "TrainRatio_mean": float(df["TrainRatio"].mean()) if "TrainRatio" in df.columns else np.nan,
        "K_mean": float(df["K"].mean()) if "K" in df.columns else np.nan,
    }

    valid_any = False

    for metric in METRICS:
        if metric in df.columns:
            vals = pd.to_numeric(df[metric], errors="coerce").dropna().values
            out[f"{metric}_mean"] = float(np.mean(vals)) if vals.size else np.nan

            if vals.size:
                valid_any = True
        else:
            out[f"{metric}_mean"] = np.nan

    if not valid_any:
        return None

    return out


def load_folder(folder: Path, pattern: str, version_label: str):
    csv_files = sorted(folder.glob(pattern))

    if not csv_files:
        print(f"[WARN] No files found in {folder.resolve()} matching {pattern}")
        return []

    rows = []

    for p in csv_files:
        rec = load_one_csv(p, version_label)

        if rec is not None:
            rows.append(rec)

    print(f"[INFO] Loaded {len(rows)} valid files for {version_label}")

    return rows


def apply_node_filter(df_all: pd.DataFrame) -> pd.DataFrame:
    if NODES_THRESHOLD is None:
        return df_all.copy()

    if not FOODWEB_CSV.exists():
        print(f"[WARN] FOODWEB_CSV not found: {FOODWEB_CSV}. Node filter skipped.")
        return df_all.copy()

    meta = pd.read_csv(FOODWEB_CSV)
    meta.columns = [c.strip() for c in meta.columns]

    fw_col = next((c for c in meta.columns if c.lower() == "foodweb"), None)
    nodes_col = next((c for c in meta.columns if c.lower() == "nodes"), None)

    if fw_col is None or nodes_col is None:
        print("[WARN] Could not find Foodweb and Nodes columns. Node filter skipped.")
        return df_all.copy()

    meta["FW_KEY"] = meta[fw_col].apply(normalize_fw_label)
    meta[nodes_col] = pd.to_numeric(meta[nodes_col], errors="coerce")

    allowed_keys = set(meta.loc[meta[nodes_col] >= NODES_THRESHOLD, "FW_KEY"])

    before = df_all["FW_KEY"].nunique()
    df_out = df_all[df_all["FW_KEY"].isin(allowed_keys)].copy()
    after = df_out["FW_KEY"].nunique()

    print(f"[INFO] Node filter: Nodes >= {NODES_THRESHOLD}")
    print(f"[INFO] Food webs before node filter: {before}")
    print(f"[INFO] Food webs after node filter: {after}")

    return df_out


def keep_complete_cases(df_all: pd.DataFrame) -> pd.DataFrame:
    if not COMPARE_ONLY_COMPLETE_CASES:
        return df_all.copy()

    required_versions = set(VERSION_ORDER)

    versions_by_fw = df_all.groupby("FW_KEY")["Version"].apply(lambda x: set(x))

    complete_keys = [
        fw for fw, versions in versions_by_fw.items()
        if required_versions.issubset(versions)
    ]

    print(f"[INFO] Complete-case food webs available in all versions: {len(complete_keys)}")

    if not complete_keys:
        raise RuntimeError(
            "No complete-case food webs found across all requested versions. "
            "Set COMPARE_ONLY_COMPLETE_CASES = False if you want to compare all available results, "
            "but this is not recommended for a fair SEAL vs WLNM comparison."
        )

    return df_all[df_all["FW_KEY"].isin(complete_keys)].copy()


def audit_seal_train_ratios(folder: Path, pattern: str):
    """
    Diagnostic tool to see why SEAL files are skipped.
    """
    print("\n[SEAL TRAIN RATIO AUDIT]")

    seal_files = sorted(folder.glob(pattern))

    if not seal_files:
        print(f"No SEAL files found in {folder.resolve()} matching {pattern}")
        return pd.DataFrame()

    audit_rows = []

    for p in seal_files:
        try:
            tmp = pd.read_csv(p)

            if "TrainRatio" not in tmp.columns:
                audit_rows.append({
                    "file": p.name,
                    "n_rows": len(tmp),
                    "min_train_ratio": np.nan,
                    "max_train_ratio": np.nan,
                    "mean_train_ratio": np.nan,
                    "has_target_ratio": False,
                    "reason": "Missing TrainRatio column",
                })
                continue

            train_vals = pd.to_numeric(tmp["TrainRatio"], errors="coerce").dropna()

            if train_vals.empty:
                audit_rows.append({
                    "file": p.name,
                    "n_rows": len(tmp),
                    "min_train_ratio": np.nan,
                    "max_train_ratio": np.nan,
                    "mean_train_ratio": np.nan,
                    "has_target_ratio": False,
                    "reason": "No numeric TrainRatio values",
                })
                continue

            has_target = bool(
                (np.abs(train_vals - TRAIN_RATIO_TARGET) <= TRAIN_RATIO_TOL).any()
            )

            audit_rows.append({
                "file": p.name,
                "n_rows": len(tmp),
                "min_train_ratio": float(train_vals.min()),
                "max_train_ratio": float(train_vals.max()),
                "mean_train_ratio": float(train_vals.mean()),
                "has_target_ratio": has_target,
                "reason": "" if has_target else f"No TrainRatio near {TRAIN_RATIO_TARGET} ± {TRAIN_RATIO_TOL}",
            })

        except Exception as e:
            audit_rows.append({
                "file": p.name,
                "n_rows": np.nan,
                "min_train_ratio": np.nan,
                "max_train_ratio": np.nan,
                "mean_train_ratio": np.nan,
                "has_target_ratio": False,
                "reason": str(e),
            })

    audit_df = pd.DataFrame(audit_rows)

    print("\nSEAL files by target-ratio availability:")
    print(audit_df["has_target_ratio"].value_counts(dropna=False).to_string())

    skipped = audit_df.loc[~audit_df["has_target_ratio"]].copy()

    print(f"\nSkipped SEAL files: {len(skipped)}")

    if not skipped.empty:
        print(
            skipped[
                ["file", "n_rows", "min_train_ratio", "max_train_ratio", "mean_train_ratio", "reason"]
            ]
            .head(MAX_AUDIT_SKIPPED_PRINT)
            .to_string(index=False)
        )

        if len(skipped) > MAX_AUDIT_SKIPPED_PRINT:
            print(f"... {len(skipped) - MAX_AUDIT_SKIPPED_PRINT} additional skipped files not shown.")

    return audit_df


# =========================
# Custom legend handler
# =========================

class HandlerMiniBoxplot(HandlerBase):
    def create_artists(
        self, legend, orig_handle,
        xdescent, ydescent, width, height, fontsize, trans
    ):
        facecolor = orig_handle["facecolor"]
        edgecolor = orig_handle.get("edgecolor", "black")
        linewidth = orig_handle.get("linewidth", 0.6)

        cx = xdescent + 0.5 * width
        cy = ydescent + 0.5 * height - 0.06 * height

        box_w = 0.8 * width
        box_h = 0.8 * height
        box_x = cx - box_w / 2
        box_y = cy - box_h / 2

        whisk_top_y = box_y + box_h + 0.18 * height
        whisk_bot_y = box_y - 0.18 * height

        box = Rectangle(
            (box_x, box_y),
            box_w,
            box_h,
            facecolor=facecolor,
            edgecolor=edgecolor,
            linewidth=linewidth,
            transform=trans,
        )

        median = Line2D(
            [box_x, box_x + box_w],
            [cy, cy],
            color=edgecolor,
            linewidth=linewidth,
            transform=trans,
        )

        whisker_top = Line2D(
            [cx, cx],
            [box_y + box_h, whisk_top_y],
            color=edgecolor,
            linewidth=linewidth,
            transform=trans,
        )

        whisker_bottom = Line2D(
            [cx, cx],
            [whisk_bot_y, box_y],
            color=edgecolor,
            linewidth=linewidth,
            transform=trans,
        )

        return [box, median, whisker_top, whisker_bottom]


# =========================
# Load data
# =========================

if RUN_SEAL_AUDIT:
    seal_audit_df = audit_seal_train_ratios(RESULTS_DIR_SEAL, GLOB_PATTERN_SEAL)

rows = []

rows += load_folder(RESULTS_DIR_ORIGINAL, GLOB_PATTERN_ORIGINAL, "Original")
rows += load_folder(RESULTS_DIR_DIRECTED, GLOB_PATTERN_DIRECTED, "Directed")
rows += load_folder(RESULTS_DIR_DIR_NEG, GLOB_PATTERN_DIR_NEG, "Directed-Negative")
rows += load_folder(RESULTS_DIR_SEAL, GLOB_PATTERN_SEAL, "SEAL")

df_all = pd.DataFrame(rows)

if df_all.empty:
    raise RuntimeError(
        "No valid rows loaded. Check paths, file patterns, and TrainRatio filter."
    )

print("\n[INFO] Raw loaded food webs by version:")
print(df_all.groupby("Version")["FW_KEY"].nunique().reindex(VERSION_ORDER).to_string())

df_all = apply_node_filter(df_all)
df_all = keep_complete_cases(df_all)

if df_all.empty:
    raise RuntimeError("No rows left after filtering.")


# =========================
# Canonical food-web names and aggregation
# =========================

display_name_map = (
    df_all
    .groupby("FW_KEY")["Foodweb"]
    .apply(choose_display_name)
    .to_dict()
)

mean_cols = [f"{m}_mean" for m in METRICS]

df_all = (
    df_all
    .groupby(["FW_KEY", "Version"], as_index=False)[
        mean_cols + ["TrainRatio_mean", "K_mean", "n_rows"]
    ]
    .mean()
)

df_all["Foodweb"] = df_all["FW_KEY"].map(display_name_map)

df_all["Version"] = pd.Categorical(
    df_all["Version"],
    categories=VERSION_ORDER,
    ordered=True,
)

print("\nUnique food webs included after complete-case matching:")
print(df_all["FW_KEY"].nunique())

print("\nRows per version:")
print(df_all["Version"].value_counts().sort_index().to_string())

print("\nFood webs included:")
print(sorted(df_all["Foodweb"].unique()))


# =========================
# Summary statistics
# =========================

print("\nSummary statistics:")
summary = (
    df_all
    .groupby("Version", observed=False)[mean_cols]
    .agg(["count", "mean", "std", "median", "min", "max"])
)

display(summary.round(4))


# =========================
# Paired comparison: SEAL - Directed-Negative
# =========================

paired_delta_rows = []

for metric in METRICS:
    metric_col = f"{metric}_mean"

    pivot = df_all.pivot_table(
        index="FW_KEY",
        columns="Version",
        values=metric_col,
        aggfunc="mean",
        observed=False,
    )

    if "SEAL" not in pivot.columns or "Directed-Negative" not in pivot.columns:
        continue

    delta = pivot["SEAL"] - pivot["Directed-Negative"]
    delta = delta.dropna()

    if delta.empty:
        continue

    paired_delta_rows.append({
        "Metric": METRIC_LABELS[metric],
        "n": int(delta.size),
        "mean_delta_SEAL_minus_DirectedNegative": float(delta.mean()),
        "median_delta_SEAL_minus_DirectedNegative": float(delta.median()),
        "std_delta": float(delta.std()),
        "min_delta": float(delta.min()),
        "max_delta": float(delta.max()),
        "SEAL_better_count": int((delta > 0).sum()),
        "DirectedNegative_better_count": int((delta < 0).sum()),
        "equal_count": int((delta == 0).sum()),
    })

delta_summary = pd.DataFrame(paired_delta_rows)

print("\nPaired delta summary: SEAL - Directed-Negative")
display(delta_summary.round(4))


# =========================
# Plot 1: boxplots across metrics
# =========================

plot_metrics = [(f"{m}_mean", METRIC_LABELS[m]) for m in METRICS]

fig, ax = plt.subplots(figsize=(12.6, 5.0))

group_gap = 1.6
group_centers = 1.0 + np.arange(len(plot_metrics)) * group_gap

offsets = {
    "Original": -0.36,
    "Directed": -0.12,
    "Directed-Negative": 0.12,
    "SEAL": 0.36,
}

box_width = 0.22

for version in VERSION_ORDER:
    data_for_version = []
    positions_for_version = []

    for i, (metric_col, _) in enumerate(plot_metrics):
        vals = (
            df_all.loc[df_all["Version"] == version, metric_col]
            .dropna()
            .values
        )

        if len(vals) == 0:
            continue

        data_for_version.append(vals)
        positions_for_version.append(group_centers[i] + offsets[version])

    if not data_for_version:
        continue

    ax.boxplot(
        data_for_version,
        positions=positions_for_version,
        widths=box_width,
        patch_artist=True,
        manage_ticks=False,
        showfliers=True,
        boxprops=dict(
            facecolor=PALETTE[version],
            edgecolor="black",
            linewidth=0.8,
            alpha=0.95,
        ),
        medianprops=dict(color="black", linewidth=1.0),
        whiskerprops=dict(color="black", linewidth=0.8),
        capprops=dict(color="black", linewidth=0.8),
        flierprops=dict(
            marker="o",
            markersize=4.5,
            markerfacecolor="black",
            markeredgecolor="black",
            markeredgewidth=0.2,
            alpha=0.65,
        ),
    )

legend_handles = [
    {
        "facecolor": PALETTE[v],
        "edgecolor": "black",
        "linewidth": 0.8,
    }
    for v in VERSION_ORDER
]

ax.set_xlim(group_centers[0] - 0.8, group_centers[-1] + 0.8)
ax.set_ylim(0, 1.0)

ax.set_xticks(group_centers)
ax.set_xticklabels(
    [label for _, label in plot_metrics],
    rotation=0,
    ha="center",
    fontsize=11,
)

ax.set_ylabel("Metric value", fontsize=12)
ax.set_xlabel("")
ax.tick_params(axis="y", labelsize=10)
ax.yaxis.set_major_locator(MaxNLocator(6))
ax.grid(True, axis="y", linestyle="--", alpha=0.35)

n_fw = df_all["FW_KEY"].nunique()

ax.set_title(
    f"WLNM variants and SEAL comparison\n"
    f"TrainRatio ≈ {TRAIN_RATIO_TARGET}%, Nodes ≥ {NODES_THRESHOLD}, food webs = {n_fw}",
    fontsize=12,
    pad=10,
)

ax.legend(
    handles=legend_handles,
    labels=VERSION_ORDER,
    handler_map={dict: HandlerMiniBoxplot()},
    loc="lower right",
    frameon=False,
    fontsize=9,
    handlelength=1.5,
    handleheight=0.9,
    handletextpad=0.6,
    borderpad=0.2,
    labelspacing=0.45,
)

sns.despine(ax=ax)
plt.tight_layout()
plt.show()


# =========================
# Plot 2: mean performance profile
# =========================

long_df = df_all.melt(
    id_vars=["Foodweb", "FW_KEY", "Version"],
    value_vars=mean_cols,
    var_name="Metric",
    value_name="Score",
)

long_df["Metric"] = long_df["Metric"].str.replace("_mean", "", regex=False)
long_df["MetricLabel"] = long_df["Metric"].map(METRIC_LABELS)

metric_order = [METRIC_LABELS[m] for m in METRICS]

summary_profile = (
    long_df
    .dropna(subset=["Score"])
    .groupby(["Version", "MetricLabel"], observed=False)["Score"]
    .mean()
    .reset_index()
)

summary_profile["MetricLabel"] = pd.Categorical(
    summary_profile["MetricLabel"],
    categories=metric_order,
    ordered=True,
)

fig, ax = plt.subplots(figsize=(11.5, 4.8))

for version in VERSION_ORDER:
    d = summary_profile[
        summary_profile["Version"] == version
    ].sort_values("MetricLabel")

    if d.empty:
        continue

    ax.plot(
        d["MetricLabel"].astype(str),
        d["Score"],
        marker="o",
        linewidth=1.8,
        markersize=5,
        label=version,
        color=PALETTE[version],
    )

ax.set_ylim(0, 1.0)
ax.set_ylabel("Mean metric value", fontsize=12)
ax.set_xlabel("")
ax.grid(True, axis="y", linestyle="--", alpha=0.35)

ax.set_title(
    "Mean performance profile across common food webs",
    fontsize=12,
    pad=10,
)

ax.legend(frameon=False, fontsize=9, loc="lower right")

sns.despine(ax=ax)
plt.tight_layout()
plt.show()


# =========================
# Plot 3: paired delta, SEAL minus Directed-Negative
# =========================

delta_plot_data = []

for metric in METRICS:
    metric_col = f"{metric}_mean"

    pivot = df_all.pivot_table(
        index="FW_KEY",
        columns="Version",
        values=metric_col,
        aggfunc="mean",
        observed=False,
    )

    if "SEAL" not in pivot.columns or "Directed-Negative" not in pivot.columns:
        continue

    delta = pivot["SEAL"] - pivot["Directed-Negative"]
    delta = delta.dropna()

    for value in delta.values:
        delta_plot_data.append({
            "Metric": METRIC_LABELS[metric],
            "Delta": value,
        })

delta_plot_df = pd.DataFrame(delta_plot_data)

if not delta_plot_df.empty:
    fig, ax = plt.subplots(figsize=(10.8, 4.6))

    data_for_boxplot = [
        delta_plot_df.loc[delta_plot_df["Metric"] == label, "Delta"].dropna().values
        for label in metric_order
    ]

    positions = np.arange(1, len(metric_order) + 1)

    ax.boxplot(
        data_for_boxplot,
        positions=positions,
        widths=0.45,
        patch_artist=True,
        showfliers=True,
        boxprops=dict(
            facecolor="#d95f02",
            edgecolor="black",
            linewidth=0.8,
            alpha=0.85,
        ),
        medianprops=dict(color="black", linewidth=1.0),
        whiskerprops=dict(color="black", linewidth=0.8),
        capprops=dict(color="black", linewidth=0.8),
        flierprops=dict(
            marker="o",
            markersize=4.5,
            markerfacecolor="black",
            markeredgecolor="black",
            markeredgewidth=0.2,
            alpha=0.65,
        ),
    )

    ax.axhline(0, linestyle="--", linewidth=1.0, color="black", alpha=0.7)

    ax.set_xticks(positions)
    ax.set_xticklabels(metric_order, rotation=0, ha="center", fontsize=11)

    ax.set_ylabel("SEAL − Directed-Negative", fontsize=12)
    ax.set_xlabel("")
    ax.grid(True, axis="y", linestyle="--", alpha=0.35)

    ax.set_title(
        "Paired performance difference between SEAL and Directed-Negative WLNM",
        fontsize=12,
        pad=10,
    )

    sns.despine(ax=ax)
    plt.tight_layout()
    plt.show()
else:
    print("[WARN] Delta plot skipped because paired SEAL and Directed-Negative data were unavailable.")


# =========================
# Optional: show food-web-level comparison table
# =========================

compact_cols = ["Foodweb", "Version"] + mean_cols

df_compact = df_all[compact_cols].copy()
df_compact = df_compact.sort_values(["Foodweb", "Version"])

print("\nFood-web-level comparison table:")
display(df_compact.round(4))
```
